```python
import pandas as pd
from sentence_transformers import SentenceTransformer
```

```python
df = pd.read_csv("preprocessed_bitext_casesS.csv")
```

```python
# initializing the sentence transformers model
model = SentenceTransformer("all-MiniLM-L6-v2")
```

```python
#Generate embeddings
def compute_embedding(text):
    return model.encode(str(text)).tolist()

df['embedding'] = df['full_text'].apply(compute_embedding)
```

```python
df.to_pickle("bitext_cases_with_embeddings.pkl")

#print(df[['full_text', 'embedding']].head())
```
